```python
#Exercício-1
Cor = "Azul"

match Cor:
  case "Azul":
    print("Cor Azul")
  case _:
    print("Outra Cor")
```

```python
#Exercício-2
numero = "Dois"
match numero:
  case "Um":
     print("Você escolheu o número Um")
  case "Dois":
     print("Você escolheu o número Dois")
  case "Três":
     print("Você escolheu o número três")
  case _:
    print("Opção Inválida")
```

```python
#Exercício-3
Bebidas = 2
match Bebidas:
  case 1:
     print("Você escolheu Água")
  case 2:
     print("Você escolheu Suco")
  case 3:
     print("Você escolheu Refrigerante")
  case _:
    print("Opção Inválida")
```

```python
#Execício-4
dia = "Sexta"
match dia:
  case "Segunda":
     print("Dia de Semana!!")
  case "Terça":
    print("Dia de Semana")
  case "Quarta":
    print("Dia de Semana")
  case "Quinta":
    print("Dia de Semana")
  case "Sexta":
    print("Dia de Semana")
  case "Sabádo":
    print("Fim de Semana")
  case "Domingo":
    print("Fim de Semana")
```

```python
#Exercício-5
Nota = "B"
match Nota:
  case "A":
     print("Excelente")
  case "B":
     print("Bom")
  case "C":
     print("Precisa Melhorar")
  case _:
    print("Nota Inválida")
```
