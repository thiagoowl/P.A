```python
# você foi contratado para desenvolver um sistema simples de caixa eletrônico.

# O sistema deverá funcionar atravéz de menu interativo utilizando um sistema de repetição while.

# O programa deve iniciar com um saldo inicial de R$ 1.000,00 e permitir que o usuário escolha escolha diferentes operações  até sair do sitema.

#funcionalidades:

# 1 - ver saldo;
# 2 - depositar dinheiro
# 3 - sacar dinheiro
# 4 - sair

# regras:

# ver saldo
# ao escolher essa opção, o sistema deverá mostrar o saldo atual da conta;

# depositar dinheiro
# o usuáro deverá informar um valor para o depósito.
# o valor digitado deve ser somado ao saldo

# sacar dinheiro
# o usuário deverá informar o valor do saque.
# se houver o suficiente - realizar o saque.
# se não houver saldo suficiente - mostrar mensagem de "saldo insuficiente".

# sair

# o programa deverá encerrar apenas quando o usuário escolher a opção.

saldo = 1000.00
opcao = 0

while opcao != 4:

  print("\n--- CAIXA ELETRÔNICO ---")
  print("opção - 1 - Saldo Atual")
  print("opção - 2 - Depositar Dinheiro")
  print("opção - 3 - Sacar Dinheiro")
  print("opção - 4 - Sair")
  try:
    opcao = int (input("digite a opção desejada: "))
  except:
    print("Entrada inválida. Por favor, digite um número.")
    continue

  if opcao == 1:
    print (f"Seu saldo atual é de R$ {saldo}")

  elif opcao == 2:
    try:
      deposito = float (input("Digite o valor do depósito: "))
      if deposito > 0:
        saldo += deposito
        print (f"Depósito de R$ {deposito} realizado com sucesso. Seu novo saldo é de R$ {saldo}")
      else:
        print("O valor do depósito deve ser positivo.")
    except:
      print("Entrada inválida para o depósito. Por favor, digite um número.")

  elif opcao == 3:
    try:
      saque = float (input("Digite o valor do saque: "))
      if saque > 0:
        if saque <= saldo:
          saldo -= saque
          print (f"Saque de R$ {saque} realizado com sucesso. Seu novo saldo é de R$ {saldo}")
        else:
          print ("Saldo insuficiente.")
      else:
        print("O valor do saque deve ser positivo.")
    except:
      print("Entrada inválida para o saque. Por favor, digite um número.")

  elif opcao == 4:
    print ("Saindo!")
  else:
    print ("Opção inválida, por favor digite uma opção entre 1 e 4.")
```

```python
# Jogo menu inteligente

# Criar um sitema que funcione até o usuário escolher sair.

# Menu

# 1 - Par/Impar
# 2 - Positivo/ Negativo
# 3 - Maior
# 4 - Sair

import random

while True:
    # Exibição do Menu
    print("\n" + "="*30)
    print("      JOGO MENU INTELIGENTE      ")
    print("="*30)
    print("[ 1 ] Jogar Par ou Ímpar")
    print("[ 2 ] Verificar Positivo ou Negativo")
    print("[ 3 ] Descobrir o Maior Número")
    print("[ 4 ] Sair do Jogo")
    print("="*30)

    opcao = input("Escolha uma opção (1-4): ").strip()
    print("-" * 30)

    # Opção 1: Par ou Ímpar (Jogo contra o computador)
    if opcao == "1":
        print("--- JOGO DO PAR OU ÍMPAR ---")
        escolha_usuario = input("Você quer Par [P] ou Ímpar [I]? ").strip().upper()

        if escolha_usuario not in ['P', 'I']:
            print(" Escolha inválida! Digite 'P' para Par ou 'I' para Ímpar.")
            continue

        num_usuario = int(input("Digite o seu número: "))
        num_computador = random.randint(0, 10)
        soma = num_usuario + num_computador
        resultado = "P" if soma % 2 == 0 else "I"

        print(f"-> Você jogou {num_usuario} e o computador jogou {num_computador}. Total: {soma}")
        print(f"-> Deu " + ("PAR!" if resultado == "P" else "ÍMPAR!"))

        if escolha_usuario == resultado:
            print(" Você VENCEU!")
        else:
            print(" O computador venceu!")

    # Opção 2: Positivo ou Negativo
    elif opcao == "2":
        print("--- POSITIVO OU NEGATIVO ---")
        num = float(input("Digite um número: "))
        if num > 0:
            print(f" O número {num} é POSITIVO.")
        elif num < 0:
            print(f" O número {num} é NEGATIVO.")
        else:
            print(" O número é ZERO (neutro).")

    # Opção 3: Maior Número
    elif opcao == "3":
        print("--- DESCOBRIR O MAIOR ---")
        n1 = float(input("Digite o primeiro número: "))
        n2 = float(input("Digite o segundo número: "))

        if n1 > n2:
            print(f" O primeiro número ({n1}) é MAIOR que o segundo ({n2}).")
        elif n2 > n1:
            print(f" O segundo número ({n2}) é MAIOR que o primeiro ({n1}).")
        else:
            print(" Ambos os números são iguais.")

    # Opção 4: Sair
    elif opcao == "4":
        print("Saindo do sistema... Obrigado por jogar! ")
        break

    # Validação para opções incorretas no menu
    else:
        print(" Opção inválida! Por favor, escolha um número de 1 a 4.")
```

```python
# Urna Eletrônica

# Criar um sistema de votação com 3 candidatos

# votar candidato A
# votar candidato B
# votar candidato C
# votar candidato D
# votar branco
# votar nulo
# encerrar votação

# o sistema deve repetir até encerrar a votação
# contar votos de cada candidato
# contar votos brancos e nulos
# ao encerrar, mostrar:

# total de votos
# total de votos por candidatos
# total de votos brancos
# total de votos nulos
# vencendor da eleição

# regras extras

# se ouver empate entre candidatos candidatos : mostrar eleição empatada
# se não ouver votos validos, mostrar: "nenhum candidato recebeu votos!!
candidato_1 = 0
candidato_2 = 0
candidato_3 = 0
brancos = 0
nulos = 0
total_votos = 0

print("="*40)
print("     SISTEMA DE URNA ELETRÔNICA      ")
print("="*40)

while True:
    print("\nOPÇÕES DE VOTO:")
    print("[ 1 ] Candidato 1")
    print("[ 2 ] Candidato 2")
    print("[ 3 ] Candidato 3")
    print("[ 4 ] Voto em Branco")
    print("[ 5 ] Voto Nulo")
    print("[ 0 ] Encerrar Votação (Apenas Mesário)")
    print("-" * 40)

    voto = input("Digite o número correspondente ao seu voto: ").strip()

    if voto == "1":
        candidato_1 += 1
        total_votos += 1
        print("Voto para Candidato 1 confirmado! ")
    elif voto == "2":
        candidato_2 += 1
        total_votos += 1
        print("Voto para Candidato 2 confirmado! ")
    elif voto == "3":
        candidato_3 += 1
        total_votos += 1
        print("Voto para Candidato 3 confirmado! ")
    elif voto == "4":
        brancos += 1
        total_votos += 1
        print("Voto em BRANCO confirmado! ")
    elif voto == "5":
        nulos += 1
        total_votos += 1
        print("Voto NULO confirmado! ")
    elif voto == "0":

        confirmar = input("Tem certeza que deseja encerrar a votação? [S/N]: ").strip().upper()
        if confirmar == "S":
            print("\nVotação encerrada com sucesso! Processando resultados...")
            break
    else:
        print(" Opção inválida! Digite um número de 1 a 5, ou 0 para encerrar.")

print("\n" + "="*40)
print("           RESULTADO FINAL           ")
print("="*40)
print(f"Total de votos computados: {total_votos}")
print(f"• Votos Brancos: {brancos}")
print(f"• Votos Nulos: {nulos}")
print("-" * 40)
print(f"• Candidato 1: {candidato_1} voto(s)")
print(f"• Candidato 2: {candidato_2} voto(s)")
print(f"• Candidato 3: {candidato_3} voto(s)")
print("-" * 40)
votos_validos = candidato_1 + candidato_2 + candidato_3
if votos_validos == 0:
    print(" Nenhum candidato recebeu votos válidos.")
else:
    maior_voto = max(candidato_1, candidato_2, candidato_3)
    vencedores = []
    if candidato_1 == maior_voto:
        vencedores.append("Candidato 1")
    if candidato_2 == maior_voto:
        vencedores.append("Candidato 2")
    if candidato_3 == maior_voto:
        vencedores.append("Candidato 3")

    #
    if len(vencedores) > 1:
        print(f" Houve um empate entre os candidatos: {', '.join(vencedores)}!")
    else:
        print(f" O vencedor da eleição é: {vencedores[0]}!")

```
