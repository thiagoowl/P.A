```python
# Cofre digital

# Um cofre possivel um código composto por 5 números inteiros.

# Desenvolva um programa que:

# Leia os 5 números utilizando um laço de repetição .

# Ao final informar:

# 1 - Quantos números são pares.
# 2 - Quantos números são impares.
# 3 - O maior número.
# 4 - A soma de todos os númeors.
# 5 - A média dos números.
# 6 - O menor números.
# 7 - Quantos números são maiores que a média.

numbers = []
for i in range(5):
    while True:
        try:
            num = int(input(f"Digite o {i+1}º número inteiro: "))
            numbers.append(num)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")

pares = 0
impares = 0
soma = 0
maior = numbers[0]
menor = numbers[0]

for num in numbers:
    if num % 2 == 0:
        pares += 1
    else:
        impares += 1
    soma += num
    if num > maior:
        maior = num
    if num < menor:
        menor = num

media = soma / len(numbers)

maiores_que_media = 0
for num in numbers:
    if num > media:
        maiores_que_media += 1

print(f"\nResultados:\n")
print(f"1 - Números pares: {pares}")
print(f"2 - Números ímpares: {impares}")
print(f"3 - Maior número: {maior}")
print(f"4 - Soma de todos os números: {soma}")
print(f"5 - Média dos números: {media:.2f}")
print(f"6 - Menor número: {menor}")
print(f"7 - Quantos números são maiores que a média: {maiores_que_media}")

```

```python
# Número espelho

# Solicite um número inteiro positivo

# Inverta o número.
# Informe se ele é um palíndromo.

while True:
    try:
        num_str = input("Digite um número inteiro positivo: ")
        num = int(num_str)
        if num < 0:
            print("O número deve ser positivo.")
        else:
            break
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

# Inverter o número
numero_invertido_str = num_str[::-1]
numero_invertido = int(numero_invertido_str)

print(f"Número original: {num}")
print(f"Número invertido: {numero_invertido}")

# Verificar se é um palíndromo
if num_str == numero_invertido_str:
    print(f"O número {num} é um palíndromo.")
else:
    print(f"O número {num} NÃO é um palíndromo.")
```
