```python
# Exercício 1 — Contagem
# faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

numero = int(input("digite um numero:"))

contador = 1

while contador <= numero:
  print(contador)
  contador = contador + 1
```

```python
# Exercício 2 — Contagem regressiva

# Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

# Exemplo:

# Digite um número: 5

# 5

# 4

# 3

# 2

# 1

# 0

numero = int(input("digite um numero:"))

contador = 1

while numero >= 0:
  print(numero)
  numero = numero - 1
```

```python
#Exercício 3 — Números pares

#Faça um programa que mostre todos os números pares de 0 até 20 usando while.

numero = int(input("digite um numero:"))

contador = 0

while contador <= numero:
  print(contador)
  contador = contador + 2
```

```python
#Exercício 4 — Soma dos números

#Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

#Exemplo:

#Digite um número: 5

#Soma: 15

#Porque:

#1 + 2 + 3 + 4 + 5 = 15

numero = int(input("Digite um número: "))

soma = 0
contador = 1

while contador <= numero:
    print(contador)
    soma += contador
    contador = contador + 1
print(f"Soma: {soma}")
```

```python
#Exercício 5 — Tabuada

#Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

#Exemplo:

#Digite um número: 7

#7 x 1 = 7

#7 x 2 = 14

#...

#7 x 10 = 70
numero = int(input("digite um numero:"))

contador = 1

while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador = contador  + 1
```

```python
#Exercício 6 — Senha

#Crie um programa que peça uma senha ao usuário.

#Enquanto a senha estiver incorreta, mostre:

#Senha incorreta!

#Quando o usuário digitar a senha correta, mostre:

#Acesso permitido!

#Considere que a senha correta seja "1234".

senha_correta = "1234"

senha = input("Digite a senha: ")

while senha != senha_correta:
    print("Senha incorreta!")
    senha = input("Digite a senha novamente: ")

print("Acesso permitido!")
```

```python
#Exercício 7 — Somando até digitar zero

#Faça um programa que peça vários números ao usuário.

#O programa deve continuar pedindo números até que o usuário digite 0.

#Ao final, mostre a soma de todos os números digitados.

#Exemplo:

#Digite um número: 10

#Digite um número: 5

#Digite um número: 8

#Digite um número: 0

#Soma: 23


soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número: "))

print(f"Soma: {soma}")
```

```python
#Exercício 8 — Maior número

#Peça números ao usuário até que ele digite 0.

#Ao final, mostre qual foi o maior número digitado.

#Exemplo:

#Digite um número: 15

#Digite um número: 8

#Digite um número: 32

#Digite um número: 20

#Digite um número: 0


#Maior número: 32

numero = int(input("Digite um número: "))

maior = numero

while numero != 0:
    if numero > maior:
        maior = numero

    numero = int(input("Digite um número: "))

if maior != 0:
    print(f"Maior número: {maior}")
else:
    print("Nenhum número válido foi digitado.")
```

```python
#Exercício 9 — Média das notas

#Faça um programa que peça 5 notas de um aluno usando while.

#Ao final, calcule e mostre a média das notas.

#Exemplo:

#Digite a nota: 7

#Digite a nota: 8

#Digite a nota: 6

#Digite a nota: 9

#Digite a nota: 10

#Média: 8.0

soma = 0
contador = 1

while contador <= 5:
    nota = float(input("Digite a nota: "))
    soma += nota
    contador += 1

media = soma / 5

print(f"Média: {media:.1f}")
```

```python
#Exercício 10 — Caixa eletrônico

#Faça um programa que comece com um saldo de R$ 1000,00.

#Mostre um menu:

#1 - Depositar

#2 - Sacar

#3 - Ver saldo

#4 - Sair
saldo = 1000.0
opcao = 0

while opcao != 4:
    print("\n--- CAIXA ELETRÔNICO ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor = float(input("Digite o valor para depósito: R$ "))
        if valor > 0:
            saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
        else:
            print("Valor inválido para depósito.")

    elif opcao == 2:
        valor = float(input("Digite o valor para saque: R$ "))
        if valor > 0 and valor <= saldo:
            saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso!")
        elif valor > saldo:
            print("Saldo insuficiente!")
        else:
            print("Valor inválido para saque.")

    elif opcao == 3:
        print(f"Seu saldo atual é: R$ {saldo:.2f}")

    elif opcao == 4:
        print("Saindo... Obrigado por utilizar o caixa eletrônico!")

    else:
        print("Opção inválida! Tente novamente.")
```
